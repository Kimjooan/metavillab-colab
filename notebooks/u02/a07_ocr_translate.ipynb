# 활동 7 · 사진 속 글자 번역하기

한국어 안내문 사진의 글자를 OCR로 읽고, 번역 모델로 한 줄씩 영어로 바꿔요.
셀을 **위에서 아래로 하나씩** 실행해요. 준비 셀, 셀 3, 셀 4는 조금 기다려야 해요.

이 노트북은 learnX 학습 화면과 같이 봐요.

```python
# [준비] 1~3분 걸려요. 기다리는 동안 학습 화면의 "이해하기"를 읽어요.
!pip install -q --progress-bar off easyocr sentencepiece sacremoses
!wget -q https://raw.githubusercontent.com/Kimjooan/metavillab-colab/main/samples/u02/sign_ko.png
import warnings, logging
warnings.filterwarnings("ignore")
logging.getLogger("easyocr").setLevel(logging.ERROR)
logging.getLogger("huggingface_hub").setLevel(logging.ERROR)
from transformers.utils import logging as hf_logging
hf_logging.set_verbosity_error()
print("준비 끝")
```

```python
# [셀 2] 사진 보기
from IPython.display import Image, display
display(Image("sign_ko.png", width=420))
```

```python
# [셀 3] 글자 읽기  ※ 1분쯤 기다려요
import easyocr
reader = easyocr.Reader(["ko", "en"], gpu=False)
results = reader.readtext("sign_ko.png")
lines = []
for box, text, conf in results:
    lines.append(text)
print("\n읽은 글자:", len(lines), "줄")
for text in lines:
    print(text)
```

```python
# [셀 4] 번역 도구 준비 (한국어를 영어로)  ※ 처음 한 번은 1분쯤 기다려요
from transformers import MarianMTModel, MarianTokenizer
model_name = "Helsinki-NLP/opus-mt-ko-en"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name, use_safetensors=True)

def translate(text):
    tokens = tokenizer([text], return_tensors="pt")
    output = model.generate(**tokens)
    return tokenizer.decode(output[0], skip_special_tokens=True)

print("번역 도구 준비 끝")
```

```python
# [셀 5] 한 줄씩 번역하기
for text in lines:
    print(text)
    print("   ", translate(text))
```

```python
# [셀 6] 내 문장 번역하기
my_sentence = "나는 파이썬으로 사진 속 글자를 읽을 수 있어요"
print(translate(my_sentence))
```
